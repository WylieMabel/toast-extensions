```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
tab = pd.read_csv("results_full_table3.csv")
tab.reset_index(inplace=True)
```

```python
t
```

```python
tab["skip_config"] = tab.apply(lambda row: f"Encoder: {row['model']}, Dataset: {row['dataset']}, Block: {row['approx_layer']}, Block Mode: {row['translator']}, MLP: {row['mlp_linearize']}, MLP Mode: {row['mlp_mode']}, Attn: {row['attn_linearize']}", axis=1)
```

```python
accuracies = tab[["skip_config", "seed", "accuracy","index"]].groupby(["skip_config"]).agg({"accuracy": ["mean", "std"], "index": ["first"]}).reset_index().sort_values(by=("index", "first"), ascending=True)
accuracies.columns = ["skip_config", "accuracy_mean", "accuracy_std", "order"]
```

```python
accuracies.to_csv("accuracies_by_skip_config_table3.csv", index=False)
```
