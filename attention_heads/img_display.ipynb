```python
from pdf2image import convert_from_path
import matplotlib.pyplot as plt
import numpy as np

# List of your 8 PDF file paths
pdf_paths = ["/Users/mabelwylie/Documents/toast-extensions/attention_heads/deitsmallCIFAR10_500.pdf",
                "/Users/mabelwylie/Documents/toast-extensions/attention_heads/deitsmallCIFAR100_500.pdf",
                "/Users/mabelwylie/Documents/toast-extensions/attention_heads/deitsmallMNIST_500.pdf",
                "/Users/mabelwylie/Documents/toast-extensions/attention_heads/deitsmallFashionMNIST_500.pdf",
                "/Users/mabelwylie/Documents/toast-extensions/attention_heads/vitbaseCIFAR10_500.pdf",
                "/Users/mabelwylie/Documents/toast-extensions/attention_heads/vitbaseCIFAR100_500.pdf",
                "/Users/mabelwylie/Documents/toast-extensions/attention_heads/vitbaseMNIST_500.pdf",
                "/Users/mabelwylie/Documents/toast-extensions/attention_heads/vitbaseFashionMNIST_500.pdf"]

fig, axes = plt.subplots(4, 2, figsize=(15, 20))
axes = axes.flatten()

for i, path in enumerate(pdf_paths):
    # Convert only the first page of the PDF to an image
    # dpi=72 is standard, increase to 200+ for higher quality
    pages = convert_from_path(path, first_page=1, last_page=1)
    
    if pages:
        # The library returns a PIL Image object
        img = pages[0]
        
        # Display it (Matplotlib handles PIL images directly, no cvtColor needed!)
        axes[i].imshow(img)
        axes[i].set_title(f"PDF {i+1}: {path.split('/')[-1]}")
        axes[i].axis('off')
    else:
        axes[i].set_title("Failed to load PDF")

plt.tight_layout()
plt.show()
```
